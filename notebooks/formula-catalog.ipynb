## 1. Diversity Indices

Diversity indices characterize the heterogeneity of knowledge sources. They can measure variety (number of categories), balance (evenness of distribution), and disparity (difference between categories).

### 1.1 Hill Numbers Framework (Hill 1973, Jost 2006, 2009)

**Hill Numbers (Diversity of Order $q$)**

$$N_q = \left( \sum_{i=1}^{n} p_i^q \right)^{\frac{1}{1-q}}$$

where:
- $p_i$ = proportional abundance of category $i$
- $q$ = order parameter (sensitivity to rare vs common categories)
- $n$ = total number of categories

**Special Cases:**

- **Order 0**: $N_0 = n$ (species richness, variety)
  - Sensitive only to presence/absence
  - Pure count of categories

- **Order 1**: $N_1 = \exp\left( -\sum_{i} p_i \ln p_i \right) = \exp(H)$
  - Exponential of Shannon entropy
  - Geometric mean of proportional abundances (weighted by themselves)
  - Limiting case as $q \to 1$

- **Order 2**: $N_2 = \frac{1}{\sum_{i} p_i^2}$
  - Inverse Simpson index
  - Arithmetic mean of proportional abundances
  - Emphasizes common categories

**Properties:**
1. **Monotonicity**: $N_q$ decreases as $q$ increases
2. **Replication Principle**: If $N$ communities have identical species abundances but no shared species, then $D(\text{pooled}) = N \cdot D(\text{single})$
3. **Partitioning Property**: $N_q(\text{within}) \times N_q(\text{between}) = N_q(\text{total})$ (multiplicative)

**Key Insight (Jost 2006, 2009)**: Raw entropies lack proper doubling behavior. Only numbers equivalents (Hill numbers) satisfy the replication principle and allow multiplicative partitioning.

---

### 1.2 Shannon Entropy Variants

**Shannon Entropy (Information-Theoretic Form)**

$$H = -\sum_{i} p_i \log_2 p_i$$

where:
- $p_i$ = proportion in category $i$
- Log base 2 gives bits of information

**Properties:**
- Maximum entropy: $H_{\max} = \log_2(N)$ where $N$ is total categories
- Can normalize as percentage of local maximum $\log(n)$ where $n \leq N$ is number of non-zero cells
- Sensitive to size effects

**Natural Log Form**

$$H = -\sum_{i} p_i \ln p_i$$

**Normalized Shannon Entropy (Pielou Evenness)**

$$J' = \frac{H}{H_{\max}} = \frac{\ln N_1}{\ln N_0}$$

**Warning**: Does NOT satisfy doubling criterion (Hill 1973). Hill evenness $E_{a,b}$ is superior.

**Conversion to True Diversity**

$$N_1 = \exp(H)$$

**Relationship to Simpson**:

$$\exp(H) \approx \frac{1}{1 - I_{\text{Simpson}}}$$

---

### 1.3 Simpson/Herfindahl Family

**Simpson Diversity (Probability of Difference)**

$$I = \sum_{i,j(i \neq j)} p_i p_j = 1 - \sum_{i} p_i^2$$

where:
- $p_i$ = proportion in category $i$

**Interpretation**: Probability that two randomly selected items belong to different categories.

**Alternative Names**: Gini-Simpson diversity, complement of Herfindahl-Hirschman index

**Herfindahl Diversity**

$$H = 1 - \sum_{i} p_i^2$$

**Note**: Identical to Simpson diversity. Complement of Herfindahl concentration index.

**Relationship to Hill Numbers**:

$$N_2 = \frac{1}{\sum_i p_i^2} = \frac{1}{1 - I}$$

**Properties**:
- Measures variety and balance but NOT disparity
- Does not account for distance between categories
- Range: $[0, 1]$ for Simpson/Herfindahl

---

### 1.4 Rao-Stirling / Integration (Multiple Variants)

**Rao-Stirling Diversity (Disparity-Based)**

$$\Delta = \sum_{i,j} d_{ij} \, p_i p_j$$

where:
- $d_{ij}$ = distance/disparity between categories $i$ and $j$
- $p_i, p_j$ = proportions in categories $i$ and $j$
- Assumes $d_{ii} = 0$ and $d_{ij} = d_{ji}$ (symmetry)

**Alternative Names**: Integration index, disciplinary diversity, quadratic entropy

**Key Innovation**: Combines vector-based (distribution) and matrix-based (network) properties.

**Integration (Similarity-Based, Xiang et al. 2025)**

$$I = 1 - \sum_{i,j} s_{ij} \cdot p_i \cdot p_j$$

where:
- $s_{ij}$ = similarity between categories $i$ and $j$
- Relationship: $s_{ij} = 1 - d_{ij}$

**Properties**:
- Captures three aspects: variety (number of categories), balance (evenness of distribution), disparity (how different categories are)
- Values compressed (products of probabilities), low discriminating power
- Choice of distance metric matters: different metrics can give negatively correlated rankings

**Porter & Rafols 2009 Form**

$$I = \sum_{i,j} s_{ij} \, p_i p_j$$

where:
- $s_{ij}$ = cosine similarity between Subject Categories $i$ and $j$
- Summation over all cells of the SC × SC matrix

**Note**: This is the Rao-Stirling diversity index with $\alpha=1, \beta=1$.

**Rao's Quadratic Entropy (Zhang et al. 2016)**

$$D = \sum_{i,j=1, i \neq j}^{N} d_{ij} \, p_i p_j$$

**Note**: When $q=1$ in generalized diversity formula. Related to but different from Rao-Stirling (uses disparity not similarity).

---

### 1.5 Stirling Generalized Framework

**Stirling Generalized Diversity**

$$\Delta_{\alpha,\beta} = \sum_{i,j} (1-s_{ij})^{\alpha} \, (p_i p_j)^{\beta}$$

where:
- $\alpha, \beta$ = parameters for weighting disparity and balance
- $s_{ij}$ = similarity between categories
- $p_i, p_j$ = proportions

**Special Cases**:
- **Rao-Stirling**: $\alpha = \beta = 1$
- **Herfindahl**: $\alpha = \beta = 0$ (distance ignored)

**Purpose**: General heuristic for exploring diversity with different weightings of disparity and balance.

---

### 1.6 Similarity-Based Diversity (Zhang et al. 2016)

**Similarity-Based Diversity (Order 2)**

$$^2D_S = \left[ \sum_{i} p_i \left( \sum_{j} s_{ij} p_j \right) \right]^{-1} = \frac{1}{\sum_{i,j} s_{ij} p_i p_j}$$

where:
- $s_{ij}$ = similarity between categories $i$ and $j$
- $s_{ii} = 1$, $0 \leq s_{ij} = s_{ji} \leq 1$

**Generalized Form**

$$^qD_S = \left[ \sum_{i} p_i \left( \sum_{j} s_{ij} p_j \right)^{q-1} \right]^{\frac{1}{1-q}}$$

where:
- $q$ = order parameter

**Key Difference**: Uses similarity matrix $S = (s_{ij})$ directly, not disparity. Different from Rao-Stirling which uses disparity $d_{ij} = 1 - s_{ij}$.

---

### 1.7 Pattern Diversity (Goyanes 2020)

**Pattern Diversity Score**

**Symbol**: $PD$

**Type**: Multidimensional diversity

**Approach**: Based on multivariate analysis combining variety, balance, disparity

**Operationalization**: Uses author affiliations, methods, theories (for communication research)

**Note**: No single closed-form formula. Uses statistical distance measures in multidimensional space.

## 2. Coherence & Network Indicators

Coherence indicators measure how tightly connected a set of publications or knowledge domains are. They use network structure rather than categorical classifications.

### Mean Linkage Strength (Rafols & Meyer 2009)

**Symbol**: $S$

**Type**: Coherence (network density)

$$S = \text{mean of bibliographic coupling matrix (excluding diagonal)}$$

where:
- Bibliographic coupling measured by Salton's cosine

**Interpretation**: Network density measure. Indicates how tightly connected a set of publications is.

**Properties**:
- Mean degree centrality normalized by network size
- Value between 0 and 1
- Higher $S$ = more coherent (tightly coupled) field

---

### Mean Path Length (Rafols & Meyer 2009)

**Symbol**: $L$

**Type**: Coherence (network compactness)

$$L = \text{average minimum number of links to traverse network}$$

where:
- Path length computed after binarizing similarities

**Interpretation**: Describes how "spread" the network is.

**Properties**:
- Equal to mean of closeness centrality
- Lower values indicate more coherent network
- Opposite relationship to $S$: low $L$ corresponds to high $S$

---

### Betweenness Centrality (Leydesdorff & Rafols 2011)

**Symbol**: $\beta$ or $BC$

**Type**: Interdisciplinarity (network position)

$$\beta_k = \sum_{i,j} \frac{g_{ikj}}{g_{ij}} \quad \text{where } i \neq j \neq k$$

where:
- $g_{ij}$ = number of geodesics (shortest paths) between vertices $i$ and $j$
- $g_{ikj}$ = number of geodesics between $i$ and $j$ that pass through $k$

**Interpretation**: Proportion of all geodesics that include vertex $k$.

**Properties**:
- Computed on binarized matrix
- Can be computed on asymmetrical citation matrix or cosine-normalized (symmetrical) matrix
- Normalization removes size effects
- High betweenness = bridging position between clusters

---

### Cosine Similarity (Leydesdorff & Rafols 2011)

**Symbol**: $\cos(x,y)$

**Type**: Similarity measure

$$\cos(x,y) = \frac{\sum_{i} x_i y_i}{\sqrt{\left(\sum_{i} x_i^2\right) \left(\sum_{i} y_i^2\right)}}$$

where:
- $x_i, y_i$ = values in vectors $x$ and $y$

**Properties**:
- Variation of correlation
- Used to normalize citation networks
- $(1 - \cos)$ can be used as distance measure
- Non-parametric, not affected by zeros like Pearson correlation

**Usage in IDR**: Foundation for computing similarity matrices $s_{ij}$ used in Rao-Stirling diversity.

---

### Co-Citation Based Metrics (Barbuti & Ferro 2020)

**XM Interdisciplinarity Metric**

$$XM_i = \frac{s_i}{s_i - 1} \times \frac{q_i}{\sum_{j} (r_{ij} - 1)}$$

where:
- $s_i$ = number of citations to paper $i$
- $q_i$ = number of duplicated co-citation links (X-motifs)
- $r_{ij}$ = number of references in citing paper $j$
- Summation over all $s_i$ citing papers

**Interdisciplinarity (IDRI)**

$$IDRI_i = 1 - XM_i$$

**Interpretation**:
- $XM_i$ measures insertion in specific field (coherence)
- $IDRI_i$ measures interdisciplinarity
- Values in $[0,1]$
- High $XM_i$ = paper embedded in coherent field
- Low $XM_i$ (high $IDRI$) = interdisciplinary

**X-Motif Count**

$$k_i = \sum_{j} (r_{ij} - 1) - q_i$$

where:
- $k_i$ = degree in co-citation network
- $r_{ij}$ = references in citing paper $j$ (excluding focal paper)
- $q_i$ = number of non-redundant X-motifs

**X-motif**: Pattern where two citing papers share two cited papers (including focal paper $i$).

**Aggregated XM-Metric (for sets of papers)**

$$XM_{\{i_1, i_2, \ldots, i_n\}} = \frac{\sum s_{i_k}}{\sum s_{i_k} - n} \times \frac{\sum q_{i_k}}{\sum \sum (r_{i_k j} - 1)}$$

**Property**: Mediant-based aggregation. Satisfies mediant inequality: aggregated value lies strictly between individual values.

**Advantage**: Classification-free. No predefined categories needed.

## 3. Inequality/Evenness Indicators

Inequality indicators measure how unevenly distributed a set of values is. In IDR context, they characterize balance across disciplines or categories.

### Gini Coefficient (Multiple Formulations)

**Rank-Based Form (Leydesdorff & Rafols 2011)**

$$G = \frac{\sum_{i} (2i - n - 1)x_i}{n \sum_{i} x_i}$$

where:
- $n$ = number of elements
- $x_i$ = number of citations of element $i$ in ranking
- $i$ = rank position

**Range**: $[0, \frac{n-1}{n}]$
- 0 = complete equality
- $\frac{n-1}{n}$ = complete inequality

**Normalized Gini Coefficient**

$$G_N = \frac{n \cdot \left[ \sum_{i} (2i - n - 1)x_i \right]}{(n-1) \cdot \sum_{i} x_i}$$

**Range**: $[0, 1]$

**Purpose**: Normalization brings all population sizes to same maximum of 1.0. Used to compare across different sized populations.

---

**Absolute Deviation Form (Leydesdorff et al. 2019)**

$$G = \frac{\sum_{i} \sum_{j} |x_i - x_j|}{2n^2 \bar{x}}$$

where:
- $x_i$ = observed value for unit $i$
- $n$ = number of observations
- $\bar{x}$ = mean value

**Note**: Equivalent to rank-based formulation.

---

**Covariance Form (Leydesdorff et al. 2019)**

$$G = \frac{2}{n^2 \bar{x}} \sum_{i} i(x_i - \bar{x})$$

**Note**: Covariance-based formulation after rank-ordering.

---

### Hill Evenness (Hill 1973)

**Symbol**: $E_{a,b}$

**Type**: Evenness

$$E_{a,b} = \frac{N_a}{N_b}$$

where:
- $N_a, N_b$ = diversity numbers of orders $a$ and $b$

**Special Case**:

$$E_{1,0} = \frac{\exp(H)}{n}$$

**Properties**:
- When all species equally abundant, $E_{a,b} = 1$ for all $a, b$
- Satisfies "doubling criterion": merging species with doubles preserves evenness
- Superior to Pielou $J'$ which does NOT satisfy doubling criterion

---

### Pielou Evenness

**Symbol**: $J'$

**Type**: Evenness

$$J' = \frac{\ln N_1}{\ln N_0} = \frac{H}{H_{\max}}$$

where:
- $N_1$ = $\exp(H)$ (exponential Shannon)
- $N_0$ = $n$ (richness)
- $H$ = Shannon entropy
- $H_{\max}$ = $\ln(n)$

**Warning**: Does NOT satisfy doubling criterion. Hill evenness $E_{a,b}$ is mathematically superior.

## 4. Novelty & Atypicality Indicators

Novelty indicators measure unusualness of knowledge combinations. They characterize how conventional or novel a paper's references are.

### Uzzi z-Scores (Uzzi 2013)

**Journal Pair z-Score**

**Symbol**: $z$

**Type**: Novelty/Conventionality

**Method**: z-score comparing observed vs expected co-citation frequency

$$z = \frac{\text{observed frequency} - \text{expected frequency}}{\text{standard deviation}}$$

where:
- Observed: frequency journal pair appears in references
- Expected: frequency from randomized citation network (Monte Carlo)

**Interpretation**:
- $z > 0$ = conventional pairing (more common than chance)
- $z < 0$ = novel pairing (less common than chance)

**Generation**: Via Monte Carlo randomization preserving citation distributions.

---

**Median Conventionality**

**Symbol**: $\text{median}(z)$

**Type**: Paper-level conventionality

$$\text{Conventionality} = \text{median of } z\text{-scores for all journal pairs in paper's references}$$

**Empirical Finding**: >50% of papers have median $z > 64$. Fewer than 4% have median $z < 0$. Science relies heavily on conventional combinations.

---

**Tail Novelty**

**Symbol**: $\text{10th percentile}(z)$

**Type**: Paper-level novelty

$$\text{Novelty} = \text{10th percentile of } z\text{-score distribution for paper's journal pairs}$$

**Interpretation**: Captures unusual combinations in left tail.

**Empirical Finding**: Only 41% of papers have 10th percentile $z < 0$.

---

**Hit Probability (Uzzi 2013)**

**Finding**: Papers with high median conventionality AND high tail novelty have 2× hit rate.

$$P(\text{hit} \mid \text{high median } z, \text{low tail } z) \approx 9.11\%$$

vs 5% baseline.

**Definition**: "Hit" = top 5% citations.

**Pattern**: All other combinations (high-high, low-high, low-low) have lower rates.

**Interpretation**: Breakthrough work combines conventional core with novel tail.

---

### Novelty Score U (Bornmann 2019, Lee/Uzzi variant)

**Symbol**: $U$ or Novelty$_U$

**Type**: Novelty

$$U = -\ln(\text{Commonness}_{10\text{th percentile}})$$

where:

$$\text{Commonness}_{ijt} = \frac{N_{ijt} \cdot N_t}{N_{it} \cdot N_{jt}}$$

and:
- $N_{ijt}$ = number of co-citation pairs from journals $i$ and $j$ in year $t$
- $N_{it}$ = number of pairs with journal $i$ in year $t$
- $N_t$ = total pairs in year $t$
- 10th percentile taken over paper's journal pairs

**Interpretation**: Higher $U$ = more novel.

**Validation**: Against F1000Prime tags. Standard deviation increase in $U$ → 7.5% increase in "new finding" tag assignments.

---

### Novelty Score W (Bornmann 2019, Wang variant)

**Symbol**: $W$

**Type**: Novelty

$$W = \sum \left( 1 - \cos(J_i, J_j) \right)$$

where:
- Sum over new journal pairs (not co-cited before focal paper)
- $\cos(J_i, J_j)$ = cosine similarity of co-citation profiles
- Co-citation profiles from 3 years prior to focal paper

**Method**:
- Only counts pairs co-cited at least once in 3 years after publication
- Excludes 50% least-cited journals

**Validation**: Mixed results. Lower convergent validity than $U$.

---

### Atypicality (Fontana 2020)

**Symbol**: $AT$

**Type**: Novelty/Atypicality

**Method**: Based on pairwise cosine distances of references

**Interpretation**: Measures unusualness of combinations.

**Note**: Related to Uzzi's approach but uses cosine distance in reference space. Empirically validated across multiple disciplines.

## 5. Key Mathematical Properties

Mathematical properties that valid diversity measures should satisfy.

### Replication Principle (Jost 2009)

**Statement**: If $N$ communities have identical species abundances but no shared species, then:

$$D(\text{pooled communities}) = N \times D(\text{single community})$$

**Significance**: Essential property for valid diversity measures.

**Who Satisfies It**:
- Hill numbers (numbers equivalents): YES
- Raw entropies (Shannon, Simpson, etc.): NO

**Implication**: Entropies must be converted to effective numbers (via exponential or inverse) before interpretation as diversity.

---

### Partitioning Property (Jost 2009)

**Statement**: For complete partitioning:

$$N_q(\text{within}) \times N_q(\text{between}) = N_q(\text{total})$$

**Type**: Multiplicative partitioning

**Validity**: For all orders $q$ when partitioning is complete.

**Who Satisfies It**:
- Hill numbers: YES
- Raw entropies: NO (additive partitioning only, and not cleanly)

---

### Doubling Criterion (Hill 1973)

**Statement**: Merging species with their identical doubles should preserve evenness measures.

**Who Satisfies It**:
- Hill evenness $E_{a,b} = N_a / N_b$: YES
- Pielou evenness $J' = H / H_{\max}$: NO

**Implication**: Hill evenness is mathematically superior.

---

### Monotonicity of Hill Numbers (Hill 1973)

**Statement**: $N_q$ is monotonically decreasing in $q$.

$$N_0 \geq N_1 \geq N_2 \geq \ldots$$

**Interpretation**: As $q$ increases, diversity measures place more weight on abundant categories and less on rare ones.

**Always True**: For any valid distribution $\{p_i\}$.

---

### Rao-Stirling Variants Relationship

**Disparity-Based Form**:

$$D = \sum_{i,j} d_{ij} \, p_i p_j$$

**Similarity-Based Form**:

$$I = 1 - \sum_{i,j} s_{ij} \, p_i p_j$$

**Relationship**:

$$d_{ij} = 1 - s_{ij}$$

**Therefore**:

$$D = \sum_{i,j} (1 - s_{ij}) p_i p_j = \sum_{i,j} p_i p_j - \sum_{i,j} s_{ij} p_i p_j$$

Since $\sum_{i,j} p_i p_j = 1$:

$$D = 1 - \sum_{i,j} s_{ij} p_i p_j = I$$

**Conclusion**: Disparity-based and similarity-based Rao-Stirling are identical up to transformation $d_{ij} = 1 - s_{ij}$.

## 7. Taxonomy of Approaches

IDR measurement can be classified by methodological approach.

### Category-Based Approaches

**Method**: Uses predefined classification schemes (e.g., WoS Subject Categories, NSF disciplines)

**Indicators**:
- Variety $N$
- Shannon entropy $H$
- Simpson diversity $I$
- Herfindahl diversity $H$
- Rao-Stirling diversity $\Delta$
- Integration $I$
- Hill numbers $N_q$

**Advantages**:
- Straightforward interpretation
- Established categories with domain meaning
- Can incorporate disparity via similarity matrices

**Disadvantages**:
- Depends on classification choice
- Categories may not align with true knowledge structure
- Classification schemes change over time

**Dimensions Captured**:
- **Variety only**: $N$ (count)
- **Variety + Balance**: Shannon $H$, Simpson $I$, Herfindahl $H$
- **Variety + Balance + Disparity**: Rao-Stirling $\Delta$, Integration $I$, Stirling generalized

---

### Network-Based Approaches

**Method**: Uses citation/reference network structure

**Indicators**:
- Mean Linkage Strength $S$ (network density)
- Mean Path Length $L$ (network compactness)
- Betweenness Centrality $\beta$ (bridging position)

**Advantages**:
- Bottom-up, data-driven
- Captures actual knowledge flows
- No predefined categories needed (for coherence measures)

**Disadvantages**:
- Size effects (can be mitigated via normalization)
- Computationally intensive
- Interpretation less straightforward

**Relationship to Diversity**: Orthogonal perspective. Diversity measures heterogeneity, coherence measures integration.

---

### Co-Citation Based Approaches

**Method**: Uses co-citation patterns without predefined categories

**Indicators**:
- XM metric (field insertion)
- IDRI (interdisciplinarity = 1 - XM)
- X-motif counts

**Advantages**:
- Classification-free
- Mediant-based aggregation (mathematically desirable property)
- Captures field coherence from citing patterns

**Disadvantages**:
- Requires sufficient citations (forward-looking)
- Less established than category-based methods
- X-motif identification can be computationally complex

---

### Novelty-Based Approaches

**Method**: Compares observed combinations to expected/baseline

**Indicators**:
- Uzzi z-scores (median conventionality, tail novelty)
- Novelty U (commonness-based)
- Novelty W (new combinations)
- Atypicality AT (cosine distance)

**Advantages**:
- Captures unusualness, not just diversity
- Validated against expert judgments and citation impact
- Distinguishes conventional core from novel tail

**Disadvantages**:
- Requires large reference corpus for baseline
- Sensitive to time windows and corpus boundaries
- Validation results mixed across different operationalizations

**Empirical Pattern**: Breakthrough work = high conventionality (median) + high novelty (tail).

---

### Hybrid Approaches

**Method**: Combines multiple perspectives

**Examples**:
- Rao-Stirling: Category-based + network-based (uses categorical proportions $p_i$ with network-derived similarities $s_{ij}$)
- Similarity diversity $^qD_S$: Hill numbers framework + similarity matrix
- Pattern diversity $PD$: Multivariate analysis combining variety, balance, disparity

**Advantage**: Captures multiple dimensions simultaneously.

**Disadvantage**: More complex interpretation and computation.

---

### Key Relationships Between Approaches

1. **Rao-Stirling variants**: $D = \sum d_{ij} p_i p_j$ (disparity) vs $I = 1 - \sum s_{ij} p_i p_j$ (similarity)
   - Related by $d_{ij} = 1 - s_{ij}$

2. **Shannon vs Simpson**: Related through exponential transformation
   - $\exp(H) \approx 1/(1-I_{\text{Simpson}})$

3. **Herfindahl = Simpson**: $H = 1 - \sum p_i^2 = \sum_{i \neq j} p_i p_j$

4. **Betweenness (citation) vs Betweenness (cosine)**: Normalization reduces size effects

5. **Diversity vs Coherence**: Orthogonal perspectives
   - Diversity = heterogeneity of knowledge sources
   - Coherence = integration/connectedness of knowledge base

6. **Hill Numbers Framework**: Unifies variety, Shannon, Simpson under single parameterized family
   - $N_0$ = variety
   - $N_1$ = $\exp(H)$
   - $N_2$ = $1/(1-I_{\text{Simpson}})$

## 8. Computational Notes

Practical considerations for implementing these formulas.

### Data Requirements

**Category-Based Diversity**:
- Requires classification scheme (e.g., ISI Subject Categories, MeSH terms, NSF disciplines)
- For each publication: proportion vector $\{p_i\}$ across categories
- For Rao-Stirling: also requires similarity matrix $S = (s_{ij})$ between categories

**Network-Based Coherence**:
- Requires similarity/distance matrix between publications
- Common: bibliographic coupling (Salton's cosine on reference lists)
- For betweenness: requires binarization threshold

**Co-Citation Based**:
- Requires forward citations (citing papers)
- For each citing paper: reference list
- Sufficient citation counts needed for stable estimates

**Novelty Metrics**:
- Uzzi z-scores: require large reference corpus + randomization procedure
- Novelty U/W: require historical co-citation frequencies (3+ years prior)
- Commonness: requires year-specific pair frequencies

---

### Normalization Issues

**Shannon Entropy**:
- Can normalize by $\log(N_{\max})$ (maximum possible categories)
- Or by $\log(n)$ where $n$ = actual categories used (local maximum)
- Local normalization more interpretable but not comparable across different $n$

**Gini Coefficient**:
- Raw: range $[0, (n-1)/n]$
- Normalized: $G_N = n \cdot G / (n-1)$ gives range $[0, 1]$
- Normalization essential for comparing across different population sizes

**Betweenness Centrality**:
- Cosine normalization removes size effects
- Alternative: normalize by $(n-1)(n-2)/2$ (maximum possible betweenness)

**Rao-Stirling**:
- Values compressed (products of probabilities)
- Low discriminating power
- No standard normalization scheme

---

### Distance Metrics for Rao-Stirling

**Cosine-Based**: $d_{ij} = 1 - \cos(c_i, c_j)$
- $c_i$ = co-citation profile of category $i$
- Performs well empirically
- Standard in bibliometrics

**Euclidean**: On normalized vectors
- Difficult to interpret results
- Less common in practice

**Choice Matters**: Different distance metrics can give negatively correlated rankings!
- Not just a technical detail
- Substantive impact on measurement

---

### Common Pitfalls

1. **Using Raw Entropies as Diversity**: Shannon $H$, Simpson $I$ do NOT satisfy replication principle. Convert to Hill numbers first.

2. **Pielou Evenness**: Fails doubling criterion. Use Hill evenness $E_{a,b}$ instead.

3. **Size Effects in Networks**: Betweenness and other network measures affected by network size. Use normalization.

4. **Classification Dependence**: Category-based measures highly sensitive to classification scheme choice. Consider multiple schemes or classification-free methods.

5. **Temporal Windows**: Novelty metrics sensitive to reference period. Document clearly.

6. **Forward Citation Requirement**: Co-citation and some novelty metrics require sufficient citations. Not applicable to very recent papers.

---

### Software Implementation Notes

**Hill Numbers**: Straightforward in any language with basic math functions.
- Watch for numerical stability when $q \approx 1$ (use limit formula)

**Rao-Stirling**: Matrix multiplication
- For large category sets, can be memory-intensive
- Sparse matrix representations help

**Network Measures**: Use established graph libraries
- Python: NetworkX
- R: igraph
- Betweenness can be slow for large networks (use approximations)

**Novelty Metrics**: Require substantial data infrastructure
- Historical co-citation frequencies
- Randomization procedures (Monte Carlo)
- Computationally intensive

---

### Validation Considerations

**Convergent Validity**: Different measures should correlate if measuring same construct
- Empirically: Shannon, Simpson, Hill numbers correlate highly
- Rao-Stirling correlates moderately with entropy measures
- Novelty U has better convergent validity than Novelty W

**Discriminant Validity**: Measures should NOT correlate with unrelated constructs
- Diversity should be orthogonal to quality
- Coherence should be orthogonal to productivity

**Criterion Validity**: Against external benchmarks
- Expert judgments (F1000Prime for novelty)
- Citation impact (for breakthrough combinations)
- Self-reported IDR (survey data)

**Robustness**: Stability under perturbations
- Small changes in classification
- Temporal windows
- Reference set boundaries

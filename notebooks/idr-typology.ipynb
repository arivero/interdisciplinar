# Multi-, Inter-, and Transdisciplinary Research: Typology and Indicator Mapping

This notebook consolidates the conceptual distinction between multi-, inter-, and transdisciplinary research
as defined in the literature, and maps each mode to observable indicator signatures. It also documents
ANECA's Campo 0 criteria as a concrete policy implementation of this distinction.

**Sources**: OECD 1998 (via Morillo et al. 2003), Wagner et al. 2011 (Table 1), Rafols & Meyer 2009,
Aksnes et al. 2026, Molas-Gallart et al. 2014, BOE-A-2023-25537 through BOE-A-2025-26118.

**Origin**: Blackboard 5 (S06, S07 cycles), 2026-02-15.

## 1. The OECD Tripartite Typology

The most widely used schema derives from the OECD (1998), refined by Morillo, Bordons, and Gómez (2003).
It distinguishes three modes of cross-disciplinary research by **integration level**:

| Mode | Definition | Integration Level |
|------|-----------|-------------------|
| **Multidisciplinary** | Different disciplinary perspectives applied in parallel; no integration achieved | Sum of parts |
| **Interdisciplinary** | Theoretical, conceptual, or methodological synthesis across disciplines | Greater than the sum of parts |
| **Transdisciplinary** | Mutual integration of disciplinary epistemologies; may transcend existing boundaries | New framework |

Morillo et al. (2003, JASIST 54:1237): *"In multidisciplinary research, the subject under study is approached
from different angles, using different disciplinary perspectives and integration is not accomplished.
Interdisciplinary research leads to the creation of a theoretical, conceptual, and methodological identity,
so more coherent and integrated results are obtained."*

Wagner et al. (2011, Table 1) refine further:
- **Multi**: "juxtapose disciplinary/professional perspectives... no more and no less than the simple sum of its parts"
- **Inter**: "integrate separate disciplinary data, methods, tools, concepts, and theories... different from, and greater than, the sum of its parts"
- **Trans**: "comprehensive frameworks that transcend the narrow scope of disciplinary worldviews"

A fourth mode is sometimes recognized:
- **Cross-disciplinary** (Aksnes 2026, Hammarfelt 2020): referencing literature from another field without any attempt at integration.

## 2. Mapping Typology to Panel Indicators

The key insight (Rafols & Meyer 2009, via Wagner et al. 2011 §5.1) is that diversity and coherence
capture fundamentally different aspects of cross-disciplinary work:

> *"Disciplinary diversity indicates the large-scale breadth of the knowledge base (the multidisciplinary scope);
> changes from lower to higher network coherence reflect knowledge integration."*

This maps the typology directly onto the three-component panel ($\Delta$, $S$, $E$):

| Mode | $\Delta$ (diversity) | $S$ (coherence) | $E$ (cross-field effect) | Bibliometric signature |
|------|---------------------|-----------------|-------------------------|------------------------|
| **Multi** | High | Low | Low–moderate | Many fields cited, no integration |
| **Inter** | Moderate–high | High | High | Multiple fields, coherent synthesis |
| **Trans** | Variable | High | High + non-academic | Requires qualitative data |
| **Specialist** | Low | High | Low | Single field, coherent |

### Diagnostic implications

- $\Delta$ alone captures **multi** (breadth) but cannot distinguish multi from inter.
- $S$ distinguishes **inter** from **multi**: high $S$ with high $\Delta$ signals integration.
- $E$ captures cross-field **impact** (diffusion beyond home discipline).
- **Trans** is not fully capturable by bibliometric indicators alone; requires survey or qualitative data.
- Single-index measures (Rao-Stirling alone) **conflate multi with inter** — this is the fundamental limitation.

## 3. Empirical Evidence: Multi and Inter Overlap

Aksnes et al. (2026) surveyed 1,498 papers with self-reported IDR ratings:

- **42%** of papers rated as BOTH multidisciplinary AND interdisciplinary
- **23%** rated as partly both
- Multi and inter are **not mutually exclusive** in practice
- Bibliometric measures (Rao-Stirling) **rarely correlate** with self-reported IDR type

This means the typology is not a clean partition but a **spectrum**, and pure diversity measures
cannot distinguish along it. The coherence component ($S$) is essential.

## 4. IDR Modalities: Long-Range vs Short-Range

Molas-Gallart et al. (2014) propose two contrasting IDR modalities using Boschma's (2005)
proximity framework:

| Dimension | Long-range IDR | Short-range IDR |
|-----------|---------------|------------------|
| Cognitive distance | SSH–STEM (long) | Within SSH (short) |
| Knowledge integration | Integrative | Multi-perspective |
| Practice | Joint teams | Specialized teams |
| Stakeholders | Collaborative | Transfer |
| Impact type | Direct use | Information |

This orthogonal distinction (long vs short range) cross-cuts the multi/inter/trans typology.
Long-range IDR with integration = interdisciplinary; long-range without = multidisciplinary.
Short-range IDR typically appears as multidisciplinary (cognitively close perspectives).

## 5. ANECA Campo 0: A Concrete Policy Implementation

### Legal basis
- **Ley Orgánica 2/2023**, Art. 11.7: *"En todos los campos se valorarán positivamente los resultados
  de investigaciones multidisciplinares e interdisciplinares."*
- Campo 0 created in 2023 resolution (BOE-A-2023-25537), updated annually.

### Campo 0 title: "Interdisciplinar y multidisciplinar"

**Transdisciplinar is NOT mentioned** in any version (2023, 2024, 2025).

### ANECA's two modes — INPUT vs OUTPUT distinction

A crucial and underappreciated aspect of ANECA's criteria is that the inter/multi distinction
tracks **directionality** (cf. Wagner et al. 2011, fn. 10: *"common usage of multi-, inter-, and
trans-disciplinary rarely distinguish between the input and output directions of IDR"*):

| Mode | ANECA definition | Direction | Indicator mapping |
|------|-----------------|-----------|-------------------|
| **Interdisciplinar** | *"diseñadas o estructuradas aplicando perspectivas, teorías o métodos asociados con diferentes disciplinas"* | **INPUT** (how research is designed) | Reference diversity $\Delta_{\mathrm{ref}}$, coherence $S$ |
| **Multidisciplinar** | *"trayectorias respaldadas con al menos dos aportaciones en campos disciplinares diferentes"* | **OUTPUT** (where results land) | Publication field diversity (variety $N \geq 2$) |

### Consequences of the input/output distinction

- A researcher can be **inter but not multi**: uses integrative methods drawing from physics and
  biology, but publishes only in biophysics journals (input-diverse, output-concentrated).
- A researcher can be **multi but not inter**: publishes separate papers in history and in computer
  science, without any methodological integration (output-diverse, input-separated).
- Evaluation must assess **both directions** — the panel ($\Delta$, $S$, $E$) computed on references
  captures the input side; publication spread captures the output side.

### Evaluation requirements
- 5 contributions of justified relevance and impact
- At least 2 different contribution typologies (from 11 disciplinary fields)
- Must justify interdisciplinary character for each contribution
- Projects with competitive funding valued positively

### Key observation

The panel ($\Delta$, $S$, $E$) operationalizes the **input side** of what ANECA requires.
The **output side** (publication diversity) requires a separate measure.
The 5-contribution minimum + 2-typology rule is itself a crude output diversity check ($N \geq 2$).

## 6. Measurement Gap Summary

| Mode | Measurable by | Gap |
|------|--------------|-----|
| **Multi** | Rao-Stirling $\Delta$, Shannon $H$, variety $N$ | Well-captured |
| **Inter** | Coherence $S$ (Rafols-Meyer) + integration evidence | Requires network analysis |
| **Trans** | No standard bibliometric indicator | Requires qualitative/survey data |

### Implication for evaluation

An evaluation framework must **first identify the mode** (multi vs inter vs trans) before selecting
appropriate indicators. Applying diversity metrics alone conflates multi with inter and misses trans entirely.

## 8. Directionality: Input/Output Flow Patterns (2026-02-15, S06 addendum)

When the output of one area is used as input for another, the IDR literature recognizes
specific patterns depending on the degree of integration:

| Pattern | Name | Example | Integration level |
|---------|------|---------|-------------------|
| A→B one-way, no synthesis | **Sequential multidisciplinarity** (Stokols 2003, via Abramo et al. 2018) | Lab uses statistical software from CS | None |
| A→B one-way, transforms B | **Instrumental/auxiliary interdisciplinarity** (Klein 1996, 2008) | ML methods transform genomics research | Partial (one direction) |
| A↔B bidirectional exchange | **Reciprocal interdisciplinarity** | Physics ↔ mathematics mutual enrichment | Full |
| A→B measured by citation patterns | **Interdisciplinary Knowledge Flow (IKF)** (Zhou, Guns & Engels 2023) | Measured by PCTC (% citations towards category) | Measured, not classified |

### Key literature

**Abramo, D'Angelo & Zhang 2018**: *"Multidisciplinary research occurs when researchers work
independently and sequentially, each from his or her own discipline-specific perspective."*

**Klein 2008** (via Abramo): Each type of disciplinarity is *"characterized by a particular type of
knowledge integration — a particular mode of merging theories and concepts, techniques and tools,
or information and data from various fields of knowledge."*

**Zhou, Guns & Engels 2023**: Propose characterizing IKF along three dimensions:
*broadness* (how many disciplines), *intensity* (volume of flow), and *homogeneity* (evenness of flow).
This goes beyond aggregate diversity indices to capture the directionality of knowledge exchange.

**Choi & Pak 2006** (via Abramo): Multi/inter/trans form *"a continuum of increasing levels of involvement
by multiple disciplines."*

### Panel indicator signatures for flow patterns

- **Sequential multi** (A→B, no synthesis): High $\Delta_{\mathrm{ref}}$ but in segregated reference clusters → low $S$
- **Instrumental inter** (A→B, transforms B): Moderate $\Delta_{\mathrm{ref}}$ with directional coherence → moderate $S$
- **Reciprocal inter** (A↔B): High $\Delta_{\mathrm{ref}}$ with bidirectional coherence → high $S$

The standard Rao-Stirling index $\Delta$ does **not** distinguish these patterns — it measures only breadth.
The coherence indicator $S$ provides partial discrimination. A full flow analysis requires directional
citation matrices (PCTC/IKF framework), which goes beyond the standard panel but could serve as
an enrichment for institutional-level assessment.
